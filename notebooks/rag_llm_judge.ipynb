# Comprehensive Evaluation V1

This notebook contains only readable configuration for the canonical benchmark. Checkpointing, scoring, route-aware judging, reporting, and charts live in `scripts/run_comprehensive_evaluation.py` and `src.evaluation.comprehensive`. Safe defaults do not call the live API or Gemini.

```python
from pathlib import Path

# ============================================================
# Acne Advisor AI - Comprehensive Evaluation Configuration
# Edit this cell only, then run the notebook from top to bottom.
# ============================================================

API_BASE_URL = "http://127.0.0.1:8000"
RUN_LIVE_EVAL = False
RUN_LLM_JUDGE = False
RESUME = False
QUESTION_LIMIT = 300
REQUEST_TIMEOUT_SECONDS = 180

DATASET_PATH = Path("notebooks/eval_data/acne_rag_eval_comprehensive_v1.jsonl")
REPORT_ROOT = Path("reports/evaluation")
EVALUATION_PROFILE = "comprehensive_v1"

LIVE_EVAL_PROVIDER = "ollama"
LIVE_EVAL_MODEL = "qwen3:8b"
BYPASS_RUNTIME_CACHE_FOR_LIVE_EVAL = True

JUDGE_PROVIDER = "gemini"
JUDGE_GEMINI_MODEL = "gemini-3.1-flash-lite"
JUDGE_SAMPLE_SIZE = 300
JUDGE_SCORE_THRESHOLD = 70
JUDGE_SLEEP_SECONDS = 2.0
JUDGE_MAX_ATTEMPTS = 5

SMOKE_MODE = False
NO_PERSISTENCE = False  # Public /chat does not support this; runner fails safely if enabled.

print("Dataset:", DATASET_PATH.name)
print("Evaluation profile:", EVALUATION_PROFILE)
print("Live answer provider:", LIVE_EVAL_PROVIDER)
print("Live answer model:", LIVE_EVAL_MODEL)
print("Judge provider:", JUDGE_PROVIDER)
print("Judge model:", JUDGE_GEMINI_MODEL)
```

```python
from pathlib import Path

from src.evaluation.comprehensive import ComprehensiveRunner, EvaluationConfig

PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

runner_config = EvaluationConfig(
    dataset_path=PROJECT_ROOT / DATASET_PATH,
    report_root=PROJECT_ROOT / REPORT_ROOT,
    api_base_url=API_BASE_URL,
    question_limit=QUESTION_LIMIT,
    live_provider=LIVE_EVAL_PROVIDER,
    live_model=LIVE_EVAL_MODEL,
    judge_provider=JUDGE_PROVIDER,
    judge_model=JUDGE_GEMINI_MODEL,
    run_live=RUN_LIVE_EVAL,
    run_judge=RUN_LLM_JUDGE,
    bypass_cache=BYPASS_RUNTIME_CACHE_FOR_LIVE_EVAL,
    request_timeout_seconds=REQUEST_TIMEOUT_SECONDS,
    judge_attempts=JUDGE_MAX_ATTEMPTS,
    judge_sleep_seconds=JUDGE_SLEEP_SECONDS,
    smoke=SMOKE_MODE,
    no_persistence=NO_PERSISTENCE,
)

print("Safe default loaded. Set RUN_LIVE_EVAL and RUN_LLM_JUDGE to True only for an approved run.")
```

```python
report_dir = None
if RUN_LIVE_EVAL and RUN_LLM_JUDGE:
    report_dir = ComprehensiveRunner(runner_config, PROJECT_ROOT).run(resume=RESUME)
    print("Comprehensive evaluation report:", report_dir)
else:
    print("Live evaluation skipped by safe defaults.")
```
